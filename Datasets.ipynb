#Correlation Heatmap

````python
import seaborn as sns
import matplotlib.pyplot as plt

corr = train.dropna().corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.figure(figsize=(10, 10))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Vehicle Sales Data Correlation Heatmap")
plt.show()
````

image.png

These are the correlations we should expect to see in our synthetic data. I listed a few examples with inuition. Notice that all of the columns are correlated with a magnitued of at least .28, suggesting strong correlations. This provides a strong benchmark to compare to synthetic data.


*   **MMR very strongly correlated with Selling price**. This make sense because MMR is Manheimn Market Report which is used to determine the selling price.

*   **Year positively correlated with selling price**. Newer cars typically cost more.

*   **Year negatively correlated with odometer**. Newer cars typically haven't been driven as far.
