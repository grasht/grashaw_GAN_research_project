# Creating the datasets

The purpose of this notebook is to download and save the dataSets to google drive for easy re-use. In the process, some pre-processing will be done. In order for this to work you must download/create a kaggle.json file. This is a file that allows you to connect to Kaggle as your user. The file is simple:

```
{
  "username": "YOUR_KAGGLE_USERNAME",
  "key": "YOUR_API_KEY"
}
```
The API-Key needs to be generated in the settings section on kaggle and is specific to your account and should not be shared.
